```python
plt.figure(figsize=(10,6))

sns.countplot(data=df, x="Department")

plt.title("Employee Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)

plt.show()
```

```python
Insight 1 — Department

The employee distribution varies across departments, with DevOps department having the highest number of employees.

```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="Region")

plt.title("Employee Distribution by Region")
plt.xlabel("Region")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)

plt.show()
```

```python
Insight 2 — Region

California region contains the largest number of employees, while Texas has the smallest representation.
```

```python
plt.figure(figsize=(10,6))

sns.histplot(df["Salary"], kde=True)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")

plt.show()
```

```python
Insight 3 — Salary

Salary values show variation across employees, with some employees receiving substantially higher salaries than the majority.
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="Performance_Score")

plt.title("Employee Performance Distribution")
plt.xlabel("Performance Score")
plt.ylabel("Number of Employees")

plt.show()
```

```python
Insight 4 — Performance

The most common performance category is Good, indicating that a large proportion of employees fall into this performance level.
```

```python
salary_department = df.groupby("Department")["Salary"].mean()

salary_department
```

```python
plt.figure(figsize=(10,6))

salary_department.sort_values().plot(kind="barh")

plt.title("Average Salary by Department")
plt.xlabel("Average Salary")
plt.ylabel("Department")

plt.show()
```

```python
Insight 6 — Department salary

Sales department has the highest average salary, while Finance has the lowest average salary.
```

```python
df["Join_Date"] = pd.to_datetime(
    df["Join_Date"],
    errors="coerce"
)
```

```python
df["Join_Year"] = df["Join_Date"].dt.year
```

```python
df["Join_Year"].value_counts().sort_index()
```

```python
plt.figure(figsize=(10,5))

df["Join_Year"].value_counts().sort_index().plot(
    kind="line",
    marker="o"
)

plt.title("Employee Joining Trend by Year")
plt.xlabel("Year")
plt.ylabel("Number of Employees")

plt.show()
```
